Good thinking — Solveit's strength is incremental, exploratory dialogs where each concept builds in small runnable steps. That favors topics that can be demoed **on CPU, in 2D, or with API calls** rather than needing GPU training runs. From my list, the standouts:

Here's a breakdown of both:

## Speculative Decoding

**Core references**
- Leviathan et al., *Fast Inference from Transformers via Speculative Decoding* (ICML 2023) — the canonical paper, with the modified rejection sampling proof
- Chen et al., *Accelerating Large Language Model Decoding with Speculative Sampling* (DeepMind, 2023) — concurrent, same core idea
- Cai et al., *Medusa: Simple LLM Inference Acceleration Framework with Multiple Decoding Heads* (2024) — draft heads instead of a draft model
- Li et al., *EAGLE: Speculative Sampling Requires Rethinking Feature Uncertainty* (2024, plus EAGLE-2/3 follow-ups) — current state of the art, used in production serving stacks
- Deployment evidence: vLLM/TensorRT-LLM speculative decoding docs show frontier adoption

**Tutorial arc (high level)**
1. *The problem*: autoregressive decoding is memory-bound — one forward pass per token, regardless of model "confidence"
2. *Key insight*: verifying K tokens in parallel costs about the same as generating 1
3. *The magic*: the accept/reject rule that makes output distribution *exactly* match the target model — provable and empirically checkable in the dialog
4. *Build it*: implement with two small models (or even character n-gram models), watch tokens proposed/accepted/rejected live with color-coded output
5. *Analysis*: acceptance rate vs. draft quality, expected speedup formula, when it fails

## Test-Time Compute / Best-of-N

**Core references**
- Cobbe et al., *Training Verifiers to Solve Math Word Problems* (2021) — GSM8K + verifiers, the origin of best-of-N with a reward model
- Lightman et al., *Let's Verify Step by Step* (2023) — process vs. outcome reward models
- Snell et al., *Scaling LLM Test-Time Compute Optimally Can Be More Effective Than Scaling Model Parameters* (2024) — the paper that framed the "test-time scaling" paradigm
- Brown et al., *Large Language Monkeys: Scaling Inference Compute with Repeated Sampling* (2024) — coverage scaling laws for repeated sampling
- OpenAI o1 / DeepSeek-R1 (2024/2025) — reasoning models as the paradigm's headline result

**Tutorial arc (high level)**
1. *The idea*: for a fixed model, spend more compute at inference to get better answers
2. *Simplest form*: sample N answers to math problems via API, majority-vote (self-consistency), plot accuracy vs. N
3. *Add a verifier*: programmatic checking vs. LLM-as-judge vs. reward model; show best-of-N beats majority vote
4. *The economics*: log-scaling of coverage, when a small model + N samples beats a big model with 1
5. *Connect to frontier*: how this leads to o1/R1-style reasoning training (RLVR as "amortized" test-time search)

**Quick comparison for your decision**: speculative decoding has the more self-contained "aha" (the lossless proof) and runs fully locally; best-of-N is cheaper to build (mostly API calls + plots) and connects more directly to the reasoning-model zeitgeist, but its insight is more empirical than conceptual.

Two good follow-ups that extend the core idea in different directions:

- **Miao et al., *SpecInfer: Accelerating Large Language Model Serving with Tree-based Speculative Inference and Verification* (ASPLOS 2024)** — generalizes the draft from a single sequence to a *token tree*, verifying many candidate continuations in one pass. This tree-attention idea is what Medusa, EAGLE-2, and most production systems build on.

- **Zhang et al., *Draft & Verify: Lossless Large Language Model Acceleration via Self-Speculative Decoding* (ACL 2024)** — removes the separate draft model entirely: the model drafts with some of its own layers skipped, then verifies with the full network. Nice for teaching because it shows the draft/verify roles are conceptual, not tied to two distinct models.

These two also give your tutorial a natural "extensions" section: sequence → tree, and two models → one model.
