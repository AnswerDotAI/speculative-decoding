## Has this existed before? Yes — NeurIPS 2025

The 2026 Education Track follows on from NeurIPS 2025, which ran a [*](https://neurips.cc/Conferences/2025/CallForEducationMaterials "NeurIPS 2025 Call For Education Materials")One-Day Event on AI Education inviting short, engaging educational materials introducing AI concepts to non-expert audiences, with accepted materials featured in an AI Education Resource Showcase. The 2025 event had a virtual session listing all accepted submissions — examples include [*](https://neurips.cc/virtual/2025/session/129859 "Track: Education Track")"From Data to CO₂: The Hidden Climate Cost of Artificial Intelligence", "A Brief Lesson on Transformers", "Interpreting Large Language Models (LLMs) with Geometry", and "AHA Moments in AI: Classroom Conversation Starters for High School Students", plus [*](https://neurips.cc/virtual/2025/session/129859 "Track: Education Track")"XAI: Explaining How AI Makes Decisions", "The AI Scientist - Physics Informed Neural Networks (PINNs)", and a hands-on interactive guide to understanding how calculus is used to train AI models. All are browsable at the OpenReview group `NeurIPS.cc/2025/Education_Program`, and the session page is at https://neurips.cc/virtual/2025/session/129859 — this is your best source of concrete accepted examples.

**Important shift for 2026:** the 2025 call targeted AI literacy for non-experts (high schoolers, general public). The 2026 call is notably different — it wants *emerging technical concepts*. [*](https://neurips.cc/Conferences/2026/CallforEducationalResources "NeurIPS 2026 Call for Educational Resources")A concept is a technical advancement that has emerged in AI research over the past few years and is shaping work across the field today — think recent equivalents of the "reparameterization trick", and [*](https://neurips.cc/Conferences/2026/CallforEducationalResources "NeurIPS 2026 Call for Educational Resources")introductions to well-established knowledge (e.g. backpropagation, transformers, or variational inference) will not be considered. So many 2025-style submissions would *not* fit 2026's scope — the example given in the call is a visual, interactive breakdown of DPO with a notebook, slides, and exercises.

## Evaluation criteria

No explicit scoring rubric is published, but the required PDF structure effectively *is* the rubric. Your two-pager must cover: [*](https://neurips.cc/Conferences/2026/CallforEducationalResources "NeurIPS 2026 Call for Educational Resources")[*](https://neurips.cc/Conferences/2026/CallforEducationalResources "NeurIPS 2026 Call for Educational Resources")the concept being taught; leveling and prerequisite knowledge; learning objectives and outcomes — a clear, specific statement of what a learner will achieve, stated [*](https://neurips.cc/Conferences/2026/CallforEducationalResources "NeurIPS 2026 Call for Educational Resources")at different levels of engagement from remembering facts to creating original work (i.e., Bloom's taxonomy framing), plus at least three recent papers from leading venues (2022–2026) that use this concept, demonstrating it is in active frontier use. Other hard requirements from the call: materials must be **original and created specifically for this track**, self-contained (source code included even if hosted publicly), ZIP under 200MB, review is **single-blind**, and at least one author must attend in person in Sydney. Key dates: submissions close **4 September 2026**, notification **28 September 2026**. Accepted work is non-archival, curated into an official repository, and eligible for a **People's Choice Award** voted by attendees.

## What this suggests for a strong submission

1. **Pick a genuinely *emerging* concept** — something researchers reach for now (post-2022) but that isn't yet in textbooks. The DPO example in the call is the calibration point.
2. **Anchor it to ≥3 frontier papers** showing active use — this is an explicit requirement and signals your concept choice is right.
3. **State prerequisites and Bloom's-style learning objectives precisely** — the call asks for these explicitly, and vague objectives are the easiest thing for reviewers to penalize.
4. **Multi-format, self-contained artifacts** — e.g. notebook + slides + short video + exercises, all included in the ZIP.
5. **Serve the track's dual goal**: a standalone resource *and* an entry point helping attendees follow NeurIPS 2026 talks that build on the concept.

Not as a track, no. The 2025 "One-Day Event on AI Education" with its Call for Education Materials was the first NeurIPS initiative of this kind, and 2026 is the first year it's branded as a full "Education Track" in the submission calls.

There were education-*adjacent* things before that, though:

- **Tutorials** — a fixture at NeurIPS for decades, but invited/proposal-based expert sessions, not open educational submissions
- **NeurIPS 2024's High School Projects track** — an open call, but for ML research projects *by* high schoolers (on social-impact topics), not teaching materials *for* learners
- **"New in ML" / affinity workshops** — onboarding-oriented events for newcomers, but workshop formats rather than reviewed educational artifacts

So the education track lineage is genuinely short: 2025 (non-expert AI literacy focus) → 2026 (emerging-concepts focus). That's worth knowing as a submitter — the track is young, norms aren't settled, and the 2025 accepted list plus the 2026 call text are essentially the entire precedent base.
